```python
plt.figure(figsize=(10,8))
plt.subplot(411)
plt.step(df["k"], df["input"], where='post',color = 'green', label = "input") 
plt.plot(df["k"], df["input"], 'o',color = 'green', label = "input", alpha=0.3) 
plt.legend()
plt.subplot(412)
plt.step(df["k"], df["state"], where='post',color = 'red', label = "state") 
plt.plot(df["k"], df["state"], 'o',color = 'red', label = "state", alpha=0.3) 
plt.legend()
plt.subplot(413)
plt.step(df["k"], df["counter"], where='post',color = 'black', label = "counter") 
plt.plot(df["k"], df["counter"], 'o',color = 'black', label = "counter", alpha=0.3) 
plt.legend()
# plt.set_ylim([0, 10])
#plt.ylim(top=15) #ymax is your value
plt.subplot(414)
plt.step(df["k"], df["output"], where='post',color = 'red', label = "output") 
plt.plot(df["k"], df["output"], 'o',color = 'red', label = "output", alpha=0.3) 
plt.legend()
#plt.step(x, y, where='post', label='post')
plt.savefig("debouncer.jpg")
plt.show()
```
